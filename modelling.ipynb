### Import Libraries

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
```

### Load The Dataset

```python
data = pd.read_csv("data/language.csv")

print(data.head())
```

### Check the Datset

```python
# Check the number of rows and columns
print("Dataset shape:", data.shape)

# Display column names
print("Columns:", data.columns.tolist())

# Check missing values
print(data.isnull().sum())

# Count examples for each language
print(data["language"].value_counts())
```

### Seperate text and labels

```python
x = np.array(data["Text"])
y = np.array(data["language"])

print("Sample text:", x[0])
print("Sample language:", y[0])
```

### Text Vectorization

```python
cv = CountVectorizer()

X = cv.fit_transform(x)

print("Feature matrix shape:", X.shape)
print("Vocabulary size:", len(cv.vocabulary_))
```

### Split the Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

### Train the ML model

```python
model = MultinomialNB()

model.fit(X_train, y_train)

print("Model training completed.")
```

### Evaluate the model

```python
accuracy = model.score(X_test, y_test)

print("Model accuracy:", accuracy)
print("Model accuracy (%):", round(accuracy * 100, 2))
```

### Predict the language of new text

```python
user_text = input("Example input")

text_vector = cv.transform([user_text])

prediction = model.predict(text_vector)

print("Predicted language:", prediction[0])
```

```python
texts = [
    "Hello, how are you?",
    "नमस्ते, आप कैसे हैं?",
    "Hola, ¿cómo estás?",
    "Bonjour, comment allez-vous?"
]

text_vectors = cv.transform(texts)
predictions = model.predict(text_vectors)

for text, language in zip(texts, predictions):
    print(f"Text: {text}")
    print(f"Predicted language: {language}")
    print("-" * 40)
```

### Save the model

```python
import joblib
from pathlib import Path

Path("models").mkdir(exist_ok=True)

joblib.dump(model, "models/language_model.pkl")
joblib.dump(cv, "models/count_vectorizer.pkl")

print("Model and vectorizer saved successfully.")
```
